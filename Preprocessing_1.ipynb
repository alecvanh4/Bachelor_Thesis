```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data2022.csv")
df
```

```python
brfss_reduced = df[['X_MICHD', 'X_SEX', 'X_AGE80', 'GENHLTH', 'X_PHYS14D', 'CVDSTRK3', 'CHCCOPD3', 
                    'CHCKDNY2', 'DIABETE4', 'HAVARTH4', 'X_BMI5', 'DIFFWALK', 
                    'SMOKE100', 'X_IMPRACE','SLEPTIM1', 'ASTHMA3', 'ADDEPEV3', ']]
```

```python
columns = ['CHD_OR_MI', 'SEX', 'AGE', 'HEALTH', 'PHYSICAL_HEALTH','STROKE',
       'BRONCHITIS', 'KIDNEY_DISEASE', 'DIABETES', 'ARTHRITIS', 'BMI', 'DIFFICULTY_WALKING', 
           'SMOKE', 'ETHNICITY','SLEEP', 'ASTHMA', 'DEPRESSION']
```

```python
brfss_reduced.columns = columns
brfss_reduced
```

```python
brfss_reduced['HEALTH'] = brfss_reduced['HEALTH'].replace({7: '', 9: ''})
brfss_reduced['PHYSICAL_HEALTH'] = brfss_reduced['PHYSICAL_HEALTH'].replace({7: '', 9: ''})
brfss_reduced['BRONCHITIS'] = brfss_reduced['BRONCHITIS'].replace({7: '', 9: ''})
brfss_reduced['KIDNEY_DISEASE'] = brfss_reduced['KIDNEY_DISEASE'].replace({7: '', 9: ''})
brfss_reduced['DIABETES'] = brfss_reduced['DIABETES'].replace({7: '', 9: ''})
brfss_reduced['DIFFICULTY_WALKING'] = brfss_reduced['DIFFICULTY_WALKING'].replace({7: '', 9: ''})
brfss_reduced['SMOKE'] = brfss_reduced['SMOKE'].replace({7: '', 9: ''})
brfss_reduced['STROKE'] = brfss_reduced['STROKE'].replace({7: '', 9: ''})
brfss_reduced['ARTHRITIS'] = brfss_reduced['ARTHRITIS'].replace({7: '', 9: ''})
brfss_reduced['ASTHMA'] = brfss_reduced['ASTHMA'].replace({7: '', 9: ''})
brfss_reduced['DEPRESSION'] = brfss_reduced['DEPRESSION'].replace({7: '', 9: ''})
brfss_reduced['SLEEP'] = brfss_reduced['SLEEP'].replace({77: '', 99: ''})
```

```python
brfss_reduced.to_csv('brfss_cleaned.csv', index=False)
```
