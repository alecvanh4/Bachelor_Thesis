```python
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
```

```python
df = pd.read_csv("brfss_clean.csv")
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=['CHD_OR_MI'])
df = df.dropna(subset=['BRONCHITIS'])
df = df.dropna(subset=['KIDNEY_DISEASE'])
df = df.dropna(subset=['HEALTH'])
df = df.dropna(subset=['DIABETES'])
df = df.dropna(subset=['ARTHRITIS'])
df = df.dropna(subset=['STROKE'])
df = df.dropna(subset=['PHYSICAL_HEALTH'])
df = df.dropna(subset=['SLEEP'])
df = df.dropna(subset=['ASTHMA'])
df = df.dropna(subset=['DEPRESSION'])
```

```python
df.isna().sum()
```

```python
na_variables = ['BMI']
```

```python
def show_distr(df):
  print("missing values : \n{}".format(df.isna().sum()))
  boxplot = df.boxplot(column=na_variables)  
  hist = df[na_variables].hist(bins=20)
  print("\n -------------------------")
```

```python
imputer_knn = KNNImputer(n_neighbors=11, weights="uniform")
df_knnimputed = imputer_knn.fit_transform(df)
```

```python
df.columns
```

```python
df_imputed = pd.DataFrame(df_knnimputed, columns=df.columns)
```

```python
df_imputed
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='most_frequent')
df_imputed[['SMOKE', 'DIFFICULTY_WALKING']] = imputer.fit_transform(df[['SMOKE', 'DIFFICULTY_WALKING']])
```

```python
df_imputed
```

```python
df_imputed.isna().sum()
```

```python
df_imputed.to_csv('brfss.csv', index=False)
```
